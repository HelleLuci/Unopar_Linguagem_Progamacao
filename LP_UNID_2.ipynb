```python
#desafio da lista de convidados
convidados = ("alice","rodrigo", "carol", "felipe")
confirmados = ["alice", "rodrigo"] #pode ser adicionado ou retirado
```

```python
print("Convidados que ainda não confirmaram:")
for convidado in convidados:
  if convidado not in confirmados: #comparando as sequencias
    print(f"{convidado}")
```
