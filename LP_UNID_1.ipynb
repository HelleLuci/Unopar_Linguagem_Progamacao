```python
# desafio classificador de filmes, com a correção

filmes = ["Filme 1", "Filme 2", "Filme 3", "Filme 4", "Filme 5"]

#criando os prints iniciais para informar o usuário
print("Bem vindo à classificação de filmes.")
print("Você tem cinco filmes para classificar!")
print("Digite '0' a qualquer momento para sair.\n") #\n pula uma linha

# criando lanço for para iterar na lista, corrigindo variavel
for filme in filmes:
  # tentando capturar entrada como inteiro
  classificacao= int(input(f"Como você classifica o filme '{filme}', ou '0' para parar: "))

  # condicionais para controle de entradas e fluxo
  if classificacao == 0:
    print("Que pena, volte sempre.")
    break
  if classificacao < 1 or classificacao > 5:
    print("Classificação inválida")
  else:
    print(f"Você classificou o '{filme}' com {classificacao}")
    print("Obrigado")
```

```python
# criando função para definir se é par
def eh_par (numero):
  if numero % 2 == 0:
    return True
  else:
    return False

# chamando a função e criando a variavel numero para ela
numero = 75
teste = eh_par(numero)
print(teste) #fazendo o print de teste e vendo o retorno funcionar
```

```python
# criar função lambda
# lambda_exp = lambda parametro: expressão

media = 7.36666688

# recebe a variável como parametro
arredondar_media = lambda media: round(media,2)

#guardar valor de forma  organizada
media_arredondada = arredondar_media(media)
print(media_arredondada)
```

```python
# testando desafio dos numeros primos
# primo é divisivel apenas por um e ele mesmo
# teste numero 4
def eh_primo (numero):
  if numero <= 1:
    return False
  # tentando iterar sobre uma sequencia e achar os numeros
  for i in range(2,numero):
    # tentando testar a divisão de todos os itens
    if numero % i == 0:
      return False # achou um divisor a mais
    # se testar todos e não achar o 0
    return True
primo = eh_primo(50)
print(primo) # se der certo eu verei os resultados
```

```python
# tentando novamente criar um for para teste dos primos
for num in range(2, 51):
  if eh_primo(num): # reutilizando código
    print(f"O numero {num} é primo")
```
